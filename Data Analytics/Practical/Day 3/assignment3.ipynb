```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
programming_languages = ['Java',' Python', 'PHP', 'JavaScript', 'C#', 'C++']
popularity = [22.2,17.6,8.8,8,7.7,6.7]
df = pd.DataFrame({'languages' : programming_languages , 'popularity':popularity})
plt.bar(x=programming_languages, height=popularity, color='blue')
plt.minorticks_on()
plt.grid(which='major', color='red', alpha=0.7)
plt.grid(which='minor', color='red', alpha=0.1) 
plt.title("Popularity of Programming Language \n Worldwide, Oct 2017 compared to a year ago")
plt.xlabel("Languages")
plt.ylabel('Popularity')
plt.legend()
plt.tight_layout()
 
```

```python
df = pd.DataFrame({
    'a':[2,4,6,8,10],
    'b':[4,2,4,2,2],
    'c':[8,3,7,6,4],
    'd':[5,4,4,4,3],
    'e':[6,6,8,6,2]
})
df
```

```python
index= np.arange(2,12,2)
width = 1
wpbar= width / 5   

plt.bar(index - wpbar*2 ,       height=df['a'],   width=wpbar     ,color = 'blue' ,label='a')
plt.bar(index - wpbar ,       height=df['b'],     width=wpbar       ,color = 'green', label='b')
plt.bar(index,              height=df['c'], width=wpbar   ,color = 'red',label='c')
plt.bar(index + wpbar,    height=df['d'],     width=wpbar       ,color = 'cyan',label='d')
plt.bar(index + wpbar*2,        height=df['e'], width=wpbar   ,color = 'purple',label='e')

plt.xticks(rotation=90)
plt.legend()

plt.minorticks_on()
plt.grid(which='major',color='green',alpha=0.6)
plt.grid(which='minor',color='green',alpha=0.2)
```
