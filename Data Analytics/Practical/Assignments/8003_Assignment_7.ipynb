```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

# ==========================================================
# Q1. Draw a line with suitable label in x, y axis and a title
# ==========================================================
plt.figure(figsize=(5, 3))
X = range(1, 50)
Y = [value * 3 for value in X]
plt.plot(X, Y)
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Draw a line.')
plt.show()

# ==========================================================
# Q2. Draw a line using given axis values with labels and title
# ==========================================================
plt.figure(figsize=(5, 3))
x = [1, 2, 3]
y = [2, 4, 1]
plt.plot(x, y)
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Sample graph!')
plt.show()
```

```python
# ==========================================================
# Q3. Draw a line using given axis values taken from a text file
# ==========================================================
plt.figure(figsize=(5, 3))
x, y = [], []
with open('test.txt', 'r') as f:
    for line in f:
        parts = line.split()
        x.append(float(parts[0]))
        y.append(float(parts[1]))
plt.plot(x, y)
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Sample graph!')
plt.show()

# ==========================================================
# Q4. Draw line charts of the financial data of Alphabet Inc.
# ==========================================================
df = pd.read_csv('fdata.csv')
df.plot(x='Date', figsize=(8, 4))
plt.title('Financial data of Alphabet Inc.')
plt.ylabel('Value')
plt.show()
```

```python
# ==========================================================
# Q5. Plot two or more lines on same plot with legends
# ==========================================================
plt.figure(figsize=(5, 3))
x1 = [10, 20, 30]
y1 = [20, 40, 10]
plt.plot(x1, y1, label='line 1')

x2 = [10, 20, 30]
y2 = [40, 10, 30]
plt.plot(x2, y2, label='line 2')

plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Two or more lines on same plot with suitable legends')
plt.legend()
plt.show()

# ==========================================================
# Q6. Plot two or more lines with legends, different widths and colors
# ==========================================================
plt.figure(figsize=(5, 3))
plt.plot(x1, y1, color='blue', linewidth=3, label='line1-width-3')
plt.plot(x2, y2, color='red', linewidth=5, label='line2-width-5')
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Two or more lines with different widths and colors')
plt.legend()
plt.show()
```

```python
# ==========================================================
# Q7. Plot two or more lines with different styles
# ==========================================================
plt.figure(figsize=(5, 3))
plt.plot(x1, y1, color='blue', linewidth=3, linestyle='dotted', label='line1-dotted')
plt.plot(x2, y2, color='red', linewidth=5, linestyle='dashed', label='line2-dashed')
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Plot with different line styles')
plt.legend()
plt.show()

# ==========================================================
# Q8. Plot two or more lines and set the line markers
# ==========================================================
plt.figure(figsize=(5, 3))
x = [2, 6, 3, 1, 5]
y = [3, 1, 4, 3, 5]
plt.plot(x, y, color='red', linestyle='dashdot', linewidth=3,
         marker='o', markerfacecolor='blue', markersize=12)
plt.xlim(1, 8)
plt.ylim(1, 8)
plt.xlabel('x - axis')
plt.ylabel('y - axis')
plt.title('Display marker')
plt.show()
```

```python
# ==========================================================
# Q9. Display current axis limit values and set new axis values
# ==========================================================
plt.figure(figsize=(5, 3))
plt.plot([1, 2, 3], [2, 4, 1])
print("Default limits:", plt.xlim(), plt.ylim())
plt.xlim(0, 10)
plt.ylim(0, 10)
plt.title('Set new axis limits')
plt.show()

# ==========================================================
# Q10. Plot quantities which have an x and y position
# ==========================================================
plt.figure(figsize=(5, 3))
x1 = [2, 3, 5, 6, 8]
y1 = [1, 5, 10, 18, 20]
x2 = [3, 4, 6, 7, 9]
y2 = [2, 6, 11, 20, 22]
plt.plot(x1, y1, 'g*', x2, y2, 'ro')
plt.title('Plot quantities with x and y position')
plt.show()
```

```python
# ==========================================================
# Q11. Plot several lines with different format styles in one command
# ==========================================================
t = np.arange(0., 5., 0.2)
plt.figure(figsize=(5, 3))
plt.plot(t, t, 'r--', t, t**2, 'bs', t, t**3, 'g^')
plt.title('Several lines in one command')
plt.show()

# ==========================================================
# Q12. Create multiple types of charts on a single set of axes
# ==========================================================
plt.figure(figsize=(5, 3))
t = np.arange(0.0, 2.0, 0.01)
s = 1 + np.sin(2 * np.pi * t)
plt.plot(t, s, color='blue', label='Curve')
plt.plot([0.5, 1.0, 1.5], [1.5, 1.0, 0.5], 'ro', label='Quantities')
plt.title('Multiple types of charts on single axis')
plt.legend()
plt.show()
```

```python
# ==========================================================
# Q13. Display grid and draw line charts with customized blue grid
# ==========================================================
plt.figure(figsize=(7, 4))
df_close = pd.read_csv('fdata.csv', usecols=['Date', 'Close'])
plt.plot(df_close['Date'], df_close['Close'], color='red', marker='o')
plt.grid(True, linestyle='-', linewidth=0.5, color='blue')
plt.xlabel('Date')
plt.ylabel('Close Value')
plt.title('Customized Blue Grid')
plt.show()

# ==========================================================
# Q14. Grid with Major and Minor lines, but turn off ticks
# ==========================================================
fig, ax = plt.subplots(figsize=(7, 4))
ax.plot(df_close['Date'], df_close['Close'], color='red', marker='o')
ax.minorticks_on()
ax.grid(which='major', linestyle='-', linewidth='0.5', color='red')
ax.grid(which='minor', linestyle=':', linewidth='0.5', color='black')
ax.tick_params(left=False, bottom=False, labelleft=True, labelbottom=True)
plt.title('Major/Minor Grid, Ticks Turned Off')
plt.show()
```

```python
# ==========================================================
# Q15. Create multiple plots (Subplots)
# ==========================================================
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(6, 6))
ax1.plot([1, 2, 3], [2, 4, 1], 'r-')
ax1.set_title('Subplot 1')
ax2.plot([1, 2, 3], [1, 3, 2], 'b-')
ax2.set_title('Subplot 2')
plt.tight_layout()
plt.show()
```
