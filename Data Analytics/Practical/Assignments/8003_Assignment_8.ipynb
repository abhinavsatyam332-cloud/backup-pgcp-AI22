### Question 4

```python
spec_date = pd.to_datetime('2023-05-15')
day_before = spec_date - pd.Timedelta(days=1)
day_after = spec_date + pd.Timedelta(days=1)

date1 = pd.to_datetime('2023-05-10')
date2 = pd.to_datetime('2023-05-20')
days_between = pd.date_range(start=date1 + pd.Timedelta(days=1), end=date2 - pd.Timedelta(days=1)).tolist()

(day_before, day_after, days_between)
```

### Question 14

### Question 20
Write a Pandas program to subtract two timestamps of same time zone or different time zone.

```python
t1 = pd.Timestamp('2023-01-02 12:00:00', tz='UTC')
t2 = pd.Timestamp('2023-01-01 12:00:00', tz='UTC')
t3 = pd.Timestamp('2023-01-01 12:00:00', tz='US/Eastern')

diff_same_tz = t1 - t2
diff_diff_tz = t1 - t3.tz_convert('UTC')

(diff_same_tz, diff_diff_tz)
```

### Question 27
Write a Pandas program to calculate one, two, three business day(s) from a specified date. Also find the next business month end from a specific date.

```python
s_date = pd.to_datetime('2023-10-13')
b1 = s_date + pd.offsets.BusinessDay(1)
b2 = s_date + pd.offsets.BusinessDay(2)
b3 = s_date + pd.offsets.BusinessDay(3)
next_bm_end = s_date + pd.offsets.BMonthEnd(1)

(b1, b2, b3, next_bm_end)
```
