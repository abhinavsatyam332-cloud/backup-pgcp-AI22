```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

Q.1. Write a Python program to draw a line with suitable label in the x axis, y
axis and a title.

```python
x = np.array([0,10,20,30,40,50])
y= np.arange(0,140,25)
plt.plot(x,y)



plt.axis([0,50,0,160])
plt.margins(x=0,y=0) 
plt.xticks([0,10,20,30,40,50])
plt.tick_params(top=True,right=True, direction = 'in')

plt.title('Draw a line')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

Q.2 Write a Python program to draw a line using given axis values with suitable
label in the x axis , y axis and a title.

```python
x = np.array([1.0,1.5,2.0,2.5,3.0])
y= np.array([2.0, 3.0 , 4.0,2.5, 1])

plt.plot(x,y)


plt.margins(x=0, y=0)  
plt.xticks(np.arange(1.0,3.1,0.5))
plt.yticks(np.arange(1.0,4.1,0.5))

plt.tick_params(top=True,right=True, direction = 'in')
plt.xlim(1.0,3.0)
plt.ylim(1.0,4.0)

plt.title('Draw a line')
plt.xlabel('x-axis')
plt.ylabel('y-axis')
```

Q.4 Write a Python program to draw line charts of the financial data of Alphabet
Inc. between October 3, 2016 to October 7, 2016.

```python
df = pd.read_csv('fdata.csv')
df['Date'] = pd.to_datetime(df['Date'], format='%m/%d/%y')
```

```python
plt.plot(df['Date'], df[['Open','High','Low','Close']], label= ['Open','High','Low','Close'])
plt.legend()
```
