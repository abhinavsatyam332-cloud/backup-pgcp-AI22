### 49. Create a second DataFrame and concatenate it with the first one.

```python
df2 = pd.DataFrame({
    'Name': ['Fiona', 'George'],
    'Age': [23, 20],
    'Marks': [91, 64],
    'Passed': [True, True],
    'Bonus': [10, 10],
    'Department': ['CS', 'ME']
})
concatenated_df = pd.concat([df, df2], ignore_index=True)
display(concatenated_df)
```

### 50. Create two DataFrames with a common column and merge them.

```python
student_details = pd.DataFrame({
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'Miami']
})
merged_df = pd.merge(df, student_details, on='Name')
display(merged_df)
```
