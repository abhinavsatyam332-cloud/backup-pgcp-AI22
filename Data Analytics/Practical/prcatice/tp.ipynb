2. Regional Analysis (WHO Regions)

```python
# How are total Confirmed cases, Deaths, and Recovered cases distributed across the different WHO Region categories?
df.groupby('WHO Region')[['Confirmed','Deaths','Recovered']].agg('sum')
```

```python
df[['WHO Region','Confirmed','Deaths','Recovered']].groupby('WHO Region').agg('sum')
```

```python
df[['WHO Region','Deaths / 100 Cases']].groupby('WHO Region').agg(['sum'])
```

```python
# Which WHO Region has the highest overall case fatality rate (Deaths / 100 Cases)?

df[['WHO Region','Deaths / 100 Cases']].groupby('WHO Region').agg(['sum'])['Deaths / 100 Cases'].sort_values(by='sum',ascending=False).head(1)

```
