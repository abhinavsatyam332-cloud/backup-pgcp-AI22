```python
# one hot encoding of categorical columns and standard scaling of numeric columns
import pandas as pd
from sklearn.preprocessing import StandardScaler,OneHotEncoder
from sklearn.impute import SimpleImputer
```

```python
df = pd.read_csv('../datasets/titanic.csv')
df
```

```python
selected_cols = ['survived','pclass','gender','age','sibsp','parch','embarked']
df = df[selected_cols]
```

```python
# separate features and label 
X = df.drop('survived',axis = 1)
y = df['survived']
```

```python
# one - hot encode categorical features
categorical_features = ['pclass',	'gender',		'sibsp',	'parch',	'embarked']
encoder = OneHotEncoder(sparse_output=False, handle_unknown='ignore')
# sparse output = false -> return a standard numpy array, not scipy array   which is the default
# handle_unkonown  = ignore  -> if we see unknown vals during transform ignore them
```

```python
#1. fit the encode one the categorical fetature
encoded_data = encoder.fit_transform(df[categorical_features])
```

```python
#2. get the new features names optional
feature_names = encoder.get_feature_names_out(categorical_features)
```

```python
# 3. create a new dataframe with encoded data and features names
encoded_df = pd.DataFrame(encoded_data, columns=feature_names)
```

```python
# now scale the numeric age column 
# handle missing values in age before scaling

missing_values_age = X['age'].isna().any()
if missing_values_age:
    print('handling nulls...')
    imputer = SimpleImputer(strategy='mean')
    imputer.fit(X[['age']])
    X['age'] = imputer.transform(X[['age']])[:,0]
    print(X.age)
```

```python
# standard scale the age feature (after imputation)
scaler = StandardScaler() 
scaler.fit(X[['age']])
X_scaled = X.copy()  # create a copy to avoid modyfyinh original data

print(X_scaled)

```

```python
X_scaled.loc[:,'age'] = scaler.transform(X[['age']])
print(X_scaled)
```

```python
# append the scaled age to the original encoded_df dataframe  
df_merged = pd.concat([encoded_df, X_scaled['age']], axis=1)
print(df_merged.head()) # print the first few rows
```
