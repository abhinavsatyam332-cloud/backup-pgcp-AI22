```python
# Linear programming: how many chairs and tables should a factory make to maximize profit?
```

```python
import numpy as np
from pulp import LpInteger,LpMaximize, LpProblem, LpVariable, value
```

```python
# 1. Initialize the optimization model (Maximize profit)
model = LpProblem('FactoryProductionProblem', LpMaximize)
```

```python
# parameters: name , lower limit, upper limit, type
# 2. Define decision variables
chairs = LpVariable('Chairs', lowBound=0, upBound=None, cat=LpInteger)
tables = LpVariable('Tables', lowBound=0, upBound=None, cat=LpInteger)

# 3. objective function
model += 300 * chairs + 1000 * tables , 'Profit'

# # 4. Define constraints
model += 5 * chairs + 10 * tables <= 400 , 'Wood_Constraint'
model += 2 * chairs + 6 * tables <= 300 , 'Time_Constraint'
# model += chairs >=2 , "Chairs_constraint"
```

```python
# 5. Solve the model
stats = model.solve()

for v in model.variables():
    print(v.name, '=', v.varValue)

print('Total profit: ', value(model.objective))
```

```python
# 6. Print the results
print(f"Status: {stats.status_str}")
print(f"Optimal Chairs to produce: {int(value(chairs))}")
print(f"Optimal Tables to produce: {int(value(tables))}")
print(f"Maximum Profit: {value(model.objective)}")
```

```python
# car example
'''
Overview
30 working days in a month
Two cars: A and B
Resources available: 1 robot, 2 engineers, 1 detailer
The detailer works 21 days
Profit per car: Car A €30,000; Car B
€45,000
Goal: maximize profit (Z)
'''
```

```python
# 1. Initialize the optimization model (Maximize profit)
model = LpProblem('CarSalesProblem', LpMaximize)

# parameters: name , lower limit, upper limit, type
# 2. Define decision variables
carsA = LpVariable('carsA', lowBound=0, upBound=None, cat=LpInteger)
carsB = LpVariable('carsB', lowBound=0, upBound=None, cat=LpInteger)

# 3. objective function
model += 30000 * carsA + 45000 * carsB, 'Profit'

# 4. Define constraints
model += 3 * carsA + 4 * carsB <=30, 'Robot time'
model += 5 * carsA + 6 * carsB <=60, 'engineer time'
model += 1.5 * carsA + 3 * carsB <=21, 'detailer time'
model += carsA >=0 , "non neg"
model += carsB >=0 , "non neg1"

# engineers are 2 hence took  carsB <= 60
stats = model.solve()

for v in model.variables():
    print(v.name, '=', v.varValue)

print('Total profit: ', value(model.objective))
```
