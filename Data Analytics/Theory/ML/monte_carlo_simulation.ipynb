```python
import numpy as np
```

```python
# PART 1
def roll_dice():
    # simulation of rolling a dice twice min number we get is 1 , max is 6
    # add the result of the two simulation , so possible values will be:
    # (1,1) or (1,2) .. (6,6) .. sum will be between 2 and 12 
    # run multiple times to verify 
    return np.sum(np.random.randint(1,7,2))

print(roll_dice())
print(roll_dice())
print(roll_dice())
```

```python
# PART 2
''' 
Someone approaches us saying i will give you  5 dollars if you get 7 and take 1 doller if you get a number other than 7
# how do we know what will happen? 
# our own monte carlo simulation like function 

'''

def monte_carlo_simulation(runs  = 1000):
    results = np.zeros(2) # win count and loss count 
    for _ in range(runs):  # run 1000 times
        if roll_dice() == 7:
            results[0] +=1
        else:
            results[1] +=1
    return results

# Test 2 - 3 times
print(monte_carlo_simulation())
print(monte_carlo_simulation())
print(monte_carlo_simulation())# result may be in favoour of us but are they 

# Part 3: Now do it 1000 times .. . takes some time 
results = np.zeros(1000)

for i in range(1000):
    results[i] = monte_carlo_simulation()[0]
print(results)


# let us plot it 
import matplotlib.pyplot as plt
fig,ax = plt.subplots()
ax.hist(results, bins=15)
plt.show()
```

```python
# our win or loss 
print(results.mean()) # mean wins
print(results.mean()*5) # winning amounts
```

```python
print(results.mean()*4.75) # winning amounts
print(results.mean()*6) # winning amounts
```
