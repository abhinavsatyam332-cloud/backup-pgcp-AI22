```python
import seaborn as sns 
import pandas as pd
import scipy.stats as stats
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = pd.read_csv('tips.csv')
total_bill_series = data['total_bill']
```

# Q1. checking data is normally distributed or not using shapiro test ans qq plot for visualization

```python
# normality test with shapiro wilk test

shapiro_test = stats.shapiro(total_bill_series)
print('test sts (W): ',shapiro_test.statistic)
print('p-val: ',shapiro_test.pvalue)
```

```python
alpha = 0.05 # common significance level
if shapiro_test.pvalue > alpha:
    print('Data is normally distributd')
else:
    print('Data is not normally distributd')
```

```python
# qq plot
plt.figure(figsize=(25,8))
stats.probplot(total_bill_series,dist='norm',plot=plt)

plt.title('QQ plot of total bill')
plt.xlabel('Theorotical quantiles (normal distribution) ')
plt.ylabel('Ordered values (total bill)')
```
