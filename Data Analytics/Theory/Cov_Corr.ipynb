```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("weight-height.csv")
df.sample(10)
```

```python
males = df.groupby('Gender').get_group('Male')
females = df.groupby('Gender').get_group('Female')


corrm = males[['Height','Weight']].corr()
covm = males[['Height','Weight']].cov()


corrf = females[['Height','Weight']].corr()
covf = females[['Height','Weight']].cov()


corr_c = df[['Height','Weight']].corr()
cov_c = df[['Height','Weight']].cov()
```

```python
print("Male correlation:\n", corrm)
print("Male covariance:\n", covm)

print("Female correlation:\n", corrf)
print("Female covariance:\n", covf)

print("Combined correlation:\n", corr_c)
print("Combined covariance:\n", cov_c)
```

```python
plt.figure(figsize=(10,6))
plt.scatter(males['Height'], males['Weight'], color='red',alpha=0.3, label='Height vs weight for males',edgecolors="white")
plt.scatter(females['Height'], females['Weight'], color='green',alpha=0.3, label='Height vs weight for females',edgecolors="white")
plt.scatter(df['Height'], df['Weight'], color='yellow',alpha=0.3, label='Height vs weight',edgecolors="white")

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(corrm, annot=True, fmt=".2f", cmap='viridis')
plt.title('correlation heatmap for males')
plt.xticks(rotation = 45)
plt.yticks(rotation = 0)
plt.show()
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(corrf, annot=True, fmt=".2f", cmap='viridis')
plt.title('correlation heatmap for female')
plt.xticks(rotation = 45)
plt.yticks(rotation = 0)
plt.show()
```
