```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python


df = pd.read_csv('marathon_results_2017.csv')
df.head(2)


```

```python
total_males= len(df[df['M/F'] == 'M'])
total_females= len(df[df['M/F'] == 'F'])
print(total_females, total_males)
origininal_row_count = len(df)
```

```python
sample_males = int(0.1 * total_males)
sample_females = int(0.1 * total_females)
print(f"Sample needs {sample_males} males and {sample_females} females")

male_sample = df[df['M/F'] == 'M'].sample(n=sample_males, random_state=42)
female_sample = df[df['M/F'] == 'F'].sample(n=sample_females, random_state=42)

sample_df = pd.concat([male_sample,female_sample])

sample_df.head()
```

```python
print(f"No of rows in original df {len(sample_df)}")
print(f"No of male athelete in sample df {(sample_df['M/F'] =='M').sum() }")
print(f"No of female athelete in sample df {(sample_df['M/F'] =='F').sum()}")
```
