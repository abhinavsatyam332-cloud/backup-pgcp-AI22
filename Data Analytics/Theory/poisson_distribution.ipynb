```python
from scipy.stats import poisson
```

```python
x = 13
meu = 15

print(f'{poisson.pmf(x, meu):.4f}')
```

```python
# 5 accidents a month at a dangereous corner
mu = 5
for x in range(5):
    print(poisson.pmf(x, mu))           # prob of 0, 1 , . . . , 4


print("\n\n")
print(poisson.cdf(2,mu))  # 0 or 1 or 2 accidents

print(1- poisson.cdf(3,mu)) # more than 3 accidents
```

Q1. exercise : A factory produces defective bulbs at an average rate of 8 bulbs per day. find probability that

1. exactly 8 defective bulbs are produced in a day.

2. exactly 5 defective bulbs are produced in a day.

3. exactly 6 defective bulbs are produced in a day.

4. more than 10 defective bulbs are produced in a day.

5. between 6 and 9 defective bulbs, inclusive are produce in a day.

```python
mu  = 8

print(f"exactly 8 defective bulbs are produced in a day: {poisson.pmf(8,mu):.4f}")
print(f"exactly 5 defective bulbs are produced in a day: {poisson.pmf(5,mu):.4f}")
print(f"exactly 6 defective bulbs are produced in a day: {poisson.pmf(6,mu):.4f}")

print(f"more than 10 defective bulbs are produced in a day: {1 - poisson.cdf(10,mu)}")

b_6_9 = poisson.cdf(9,mu) - poisson.cdf(5,mu)
print(f"between 6 and 9 defective bilbs inclusive both: {b_6_9:.4f}")
```
