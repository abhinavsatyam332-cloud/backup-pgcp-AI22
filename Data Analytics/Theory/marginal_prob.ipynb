```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('titanic.csv')
df.head()
```

```python
total = len(df)
total_female = len(df[df['gender']=='female'])
total_female

total_male = len(df[df['gender']=='female'])
total_male 
```

```python
p_female = total_female / total
p_male = total_male / total
print(p_female, "    ",p_male)
```

```python
# marginal probability of survival
total_survived = len(df[df['survived'] == 1])
print(f"total survived: {total_survived}")

total_did_not_survive = len(df[df['survived'] == 0])
print(f"total did not survived: {total_did_not_survive}")

p_survived = total_survived / total
p_not_survived = total_did_not_survive / total

print("\n Marginal probabilities: ")
print(f"P(gender = female): {p_female:.3f}")
print(f"P(gender = male): {p_male:.3f}")
print(f"P(survived = 1): {p_survived:.3f}")
print(f"P(survived = 0): {p_not_survived:.3f}")
```

```python
# joint probability
p_female_survied = len(df[(df['gender'] =='female') & (df['survived'] == 1) ]) / total
p_female_not_survied = len(df[(df['gender'] =='female') & (df['survived'] == 0) ]) / total
p_male_survied = len(df[(df['gender'] =='male') & (df['survived'] == 1) ]) / total
p_male_not_survied = len(df[(df['gender'] =='male') & (df['survived'] == 0) ]) / total

print("\n JOint probabilities: ")
print(f"p(female, survived = 1): {p_female_survied:.3f}")
print(f"p(female, survived = 0): {p_female_not_survied:.3f}")

print(f"p(male, survived = 1): {p_male_survied:.3f}")
print(f"p(male, survived = 0): {p_male_not_survied:.3f}")
```

```python
# conditional probabilities
P_survied_given_female = p_female_survied / p_female
print(f"p(survived = 1 | gender =female): {P_survied_given_female:.3f}")

P_survied_given_male = p_male_survied / p_male
print(f"p(survived = 1 | gender =male): {P_survied_given_male:.3f}")


P_not_survied_given_female = p_female_not_survied / p_female
print(f"p(survived = 0 | gender =female): {P_not_survied_given_female:.3f}")


P_not_survied_given_male = p_male_not_survied / p_male
print(f"p(survived = 0 | gender =male): {P_not_survied_given_male:.3f}")

```
