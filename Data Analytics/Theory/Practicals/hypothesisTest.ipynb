problem:
A cola company claims that the average bottle contains 500 ml

A sample of 25 bottle has an average of 503.4 ml 

sample standard deviation = 10 ml

test at 5% significance level wether the average is greater then 500ml 


```python
from scipy.stats import norm
```

```python
mu = 500  # claimed population mean
x_bar = 503.4  # sample mean
s = 10 # sample sd
n = 25      # sample size

alpha = 0.05  # significance level
```

```python
se = s / n**0.5 # standard error
```

```python
# z-test statistic
z  = (x_bar - mu) / se
z
```

```python
# critical val
# percent point function - probability -> z-score
# if p = 0.025, z-score = 1.96
# alpha = 0.05 , alpha/2 = 0.025 
critical_val = norm.ppf(1-alpha/2)             # where does rejection zone start
critical_val
```

```python
# p-value
# survival function - give the probability to the right of a z-score
p_value= 2* norm.sf(abs(z))
```

```python
print(f'''   
population mean = {mu}
sample mean = {x_bar}
standard deviation = {s}
sample size = {n}
alpha= {alpha}
standard error = {se}
z statistics = {z}
critical value = {critical_val}
p_value = {p_value}

  ''')
```

```python
#  critival value method
if z > critical_val:
    print("Critical val method: Reject H0")
else:
    print("Critical val method: Do not Reject H0")
```

```python
#  p-value
# survival function give the prob 

if p_value > alpha:
    print("p val method: Do not Reject H0")
else:
    print("p val method: Reject H0")
```
