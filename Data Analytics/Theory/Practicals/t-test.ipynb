```python
# 1-sample T-test: Is average population tip = $3
# h0 : meu = average tip = $3
```

```python
import pandas as pd
from scipy import stats
```

```python
df = pd.read_csv('datasets/tips.csv')
our_column = df['tip']
our_column = our_column.sample(n=30, replace=False)

meu = 3 # meu

# performance one sample t-test
t_statistics, p_value = stats.ttest_1samp(our_column, meu)
```

```python
# automcatically returns 2 tailed p-value  (in manual calculation we doubles the p-value , python handles this automatically)
print("p value: ",p_value)
'''  
as we are going to compare alpha and p_val then both should compare from both the sides, as alpha should doubled or p_value should double to cover entire region.

     '''
```

```python
alpha = 0.05
if p_value < alpha:
    print('reject h0')
else:
    print('do not reject h0')
```

```python
#one sampled t-test  (one tailed)
import pandas as pd
from scipy import stats
df = pd.read_csv('datasets/tips.csv')
our_column = df['tip'].sample(n = 30, replace=False)

#hypothesis value under h0
# although we say meu = 3, python considers it as <=3 (see next part)
meu = 2

# perform one-sampleright-tailed t-test
#h0 : mu <= 3
#ha : mu > 3
result = stats.ttest_1samp(our_column, meu, alternative='greater') # greater means h0 : <=and aleternative : > 

print('mu ',meu)
print('t-statistics: ',result.statistic)
print('p-val : ',result.pvalue)
print('alpha: ',alpha)
alpha = 0.05
if result.pvalue < alpha:
    print('reject h0 (mean is significantly greater than ',meu,')')
else:
    print('do not reject h0')
```

```python
''' 

Exercise : use loan dataset. run two tailed and one tailed one sample t-test. 
Two-tailed -> h0 : avg income = 50 , h1 : average income <> 50
one-tailed -> h0 : avg income >= 50 . h1 : avg income < 50

'''
```

```python
# two tailed test
df =  pd.read_csv('datasets/loan.csv')
inc_col = df['income'].sample(n=30,replace=False)


mu = 50
result = stats.ttest_1samp(inc_col, mu,alternative='two-sided')
alpha = 0.05


print('p value: ',result.pvalue)
print('alpha: ',alpha)
if result.pvalue < alpha:
    print('reject h0')
else:
    print('do not reject h0 ')
```

```python
#one tailed test
result = stats.ttest_1samp(inc_col, mu, alternative='less') # greater less two-sided

alpha = 0.05
print('p value: ',result.pvalue)
print('alpha: ',alpha)
if result.pvalue < alpha:
    print('reject h0')
else:
    print('do not reject h0 ')
```

separate data by gender t  -test

```python
df = pd.read_csv('datasets/tips.csv')
```

```python
# Separate df by gender
male_tips = df[df['gender'] == 'Male']['tip']
female_tips = df[df['gender'] == 'Female']['tip']

male_tips = male_tips.sample(n=30, replace=False)
female_tips = female_tips.sample(n=30, replace=False)

print("Male mean tip:", male_tips.mean())
print("Female mean tip:", female_tips.mean())

# Perform two-sample t-test
t_statistic, p_value = stats.ttest_ind(male_tips, female_tips)

# Significance level (alpha)
alpha = 0.05

print('t-statistics: ',t_statistic)
print('p-value: ',p_value)

if p_value < alpha:
    print('There is a significant difference in spending score between ,male and female groups')
else:
    print('There is no significant difference in spending score between ,male and female groups')
```

```python
# exercise : is there a statistically significant difference in the avg runs scored by sachin tendulkar when playing against australia vs pakistan? (sachin_tendular.csv)

df= pd.read_csv('datasets/Sachin_Tendulkar.csv')
df
```

```python
australia_runs = df[df['Versus']=='Australia']['Runs']
pakistan_runs = df[df['Versus'] == 'Pakistan']['Runs']

# australia_runs = australia_runs.sample(n=30)
# pakistan_runs = pakistan_runs.sample(n=30)



# Perform two-sample independent t-test
t_stat, p_val = stats.ttest_ind(australia_runs, pakistan_runs)

print("Australia Mean:", australia_runs.mean())
print("Pakistan Mean:", pakistan_runs.mean())
print("p-value:", p_val)

alpha = 0.05
if p_val < alpha:
    print("Reject H0: Statistically significant difference between performance vs Australia and Pakistan.")
else:
    print("Do not reject H0: No statistically significant difference found.")
```

paired t-test: a company has launched some offer

h0: pre and post usage is the same (i.e. offer is not effective)

```python
from scipy.stats import ttest_rel
```

```python
df = pd.read_csv('datasets/cust_seg.csv')
df
usage_data = df[['pre_usage','Post_usage_1month']]
usage_data = usage_data.sample(n=30)



tstat, pval = ttest_rel(usage_data['pre_usage'], usage_data['Post_usage_1month'])

alpha = 0.5
if pval < alpha:
    print('reject null hypothesis')
else:
    print('fail to reject null hypothesis')
```
