```python
# binomial distribution
from scipy.stats import binom
```

```python
n = 5  #(no of students asked (trials))
k = 3  # number who like python (success)
p = 0.66  # prop that one student like python 
```

```python
pmf = binom.pmf(k, n, p)   # PMF : p(x=k)
cdf = binom.cdf(k, n, p)   # CDF : p(x<=k)

print(f"PMF p(x = {k})  =  {pmf:.4f}")
print(f"CDF p(x <= {k})  =  {cdf:.4f}")
```

```python
#PDF
prob = binom.cdf(3,n,p) - binom.cdf(1,n,p)
print(f"binomial probability p(2<=x<=3) = {prob:.4f}")
```

```python
#pdf another method
pmf_2 = binom.pmf(2,n,p)  # pmf: p(x=2)
pmf_3 = binom.pmf(3,n,p)  # pmf: p(x=3)
prob = pmf_2 + pmf_3
print(f"Binomial prob p(2<=x<=3) = {prob:.4f}")
```

```python
# Q1.  For my practice
# 75% of patients with a disease die of it. six patient are selected what is the probability that (a) exactly 4 recover,    (b) more that 4 recover
n = 6
k = 4
p = 0.25  # prob of patient with a disease dont die

pmf = binom.pmf(k,n,p)
cdf = binom.cdf(k,n,p)
print(f"PMF p(x = {k}) exactly 4 recover =  {pmf:.4f}")
print(f"CDF p(x <= {k})  =  {cdf:.4f}")
```

```python
prob = binom.cdf(6,n,p) - binom.cdf(4,n,p)
print(f"prob that more than 4 recover = {prob:.4f}")
```

```python
# or 
prob = binom.pmf(5,n,p) + binom.pmf(6,n,p)
print(f"prob that more than 4 recover = {prob:.4f}")
```

```python
# or
prob = 1 - binom.cdf(4,n,p)
print(f"prob that more than 4 recover = {prob:.4f}")
```

Q2  A marksman hits the target 4 times out of 5 . find probability that when he fires 4 shots : 

a. more than 2 hits

b. at least 3 are misses ( = at most 1 hit)

```python
n = 4
p = 0.8

m2h = binom.cdf(4,n,p) - binom.cdf(2,n,p)

m1h = binom.pmf(1,n,p)

print(f"More than 2 hits: {m2h:.4f}")
print(f"at least 3 are misses: {m1h:.4f}")
```

Q3. 65% of students who appear for the c-dac entrance test clear it. from a city 50 students appear. find probability that

a. nobady clears

b. more than 40 clear

```python
n = 50
p = 0.65

nob_clear = 1 - binom.pmf(2,n,p) 

m_40_clear = binom.cdf(50, n,p) - binom.cdf(40, n, p )

print(f"Nobady clears: {nob_clear:.4f}")
print(f"more than 40 clears: {m_40_clear:.4f}")
```
