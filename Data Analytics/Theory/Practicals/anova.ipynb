```python
# ANOVA: h0: parents education level has no impact on the childs math score
```

```python
from scipy.stats import f_oneway
import pandas as pd 

df = pd.read_csv('datasets/StudentsPerformance.csv')
```

```python
g1 = df['math score'][df['parental level of education'] == 'some high school']
g2 = df['math score'][df['parental level of education'] == 'high school']
g3 = df['math score'][df['parental level of education'] == 'some college']
g4 = df['math score'][df['parental level of education'] == 'associate\'s degree']
g5 = df['math score'][df['parental level of education'] == 'bachelor\'s degree']
g6 = df['math score'][df['parental level of education'] == 'master\'s degree']
print(g1)
fstat, pval = f_oneway(g1,g2,g3,g4,g5,g6)
print(f'''
fstat: {fstat}
pval: {pval}

 ''')

alpha = 0.05

if pval < alpha:
    print('reject h0')
else:
    print('do not reject h0')
```

whose total bill amounts are < $5 , $5-10, $10-15, > $15 in the tips dataset? 

(handle boundary conditions appropriately)

```python
df=pd.read_csv('datasets/tips.csv')             # if more than 2 groups take annova
df
g1 = df['tip'][df['total_bill']< 13]
g2 = df['tip'][(df['total_bill']>= 13) & (df['total_bill']< 17) ]
g3 = df['tip'][(df['total_bill']>= 17) & (df['total_bill']< 22)]
g4 = df['tip'][(df['total_bill']>= 22)]

fstat, pval = f_oneway(g1,g2,g3,g4)
alpha  = 0.05 
print("Sample sizes:", len(g1), len(g2), len(g3), len(g4))
print(f'''
fstat: {fstat}
pval: {pval:.25f}

 ''')
if pval < alpha:
    print('reject h0')
else:
    print('do not reject h0')
```
