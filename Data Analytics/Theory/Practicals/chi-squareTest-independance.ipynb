```python
import pandas as pd
from scipy.stats import chi2_contingency
```

```python
# chi-square test of independence

# there is no statistically significant dependence between cancelled / pending orders and order fulfilment. 
```

```python
df = pd.read_csv('datasets/Amazon.csv',low_memory=False)
df


cancelled_pending_orders = df[df['Status'].isin(['Cancelled','Pending'])]
contingency_table = pd.crosstab(cancelled_pending_orders['Status'], cancelled_pending_orders['Fulfilment'])
print('\n\n',contingency_table)

#perform chi-square test 
chi2, p ,dof, ex = chi2_contingency(contingency_table)

print('\n\nchi-squared statistics: ',chi2)
print('p-val', p)
print('degrees of freedom: ',dof)

print('\n\n')
print('expected values: \n',ex)

if p < 0.05:
    print('\nreject null hypothisis')
else:
    print('\nfail to reject null hypothesis')
```

```python
df = pd.read_csv('datasets/titanic.csv')
df
```

```python
print('for a. ')
contingency_table = pd.crosstab(df['gender'], df['survived'])
print(contingency_table)


#perform chi-square test 
chi2, p ,dof, ex = chi2_contingency(contingency_table)

print('\n\nchi-squared statistics: ',chi2)
print('p-val', p)
print('degrees of freedom: ',dof)

print('\n\n')
print('expected values: \n',ex)

alpha = 0.05
if p < alpha:
    print('\nReject H0: Gender and survival are NOT independent (there is a significant relationship).')
else:
    print('\nFail to reject H0: Gender and survival are independent.')
```

```python
print('for b. ')
contingency_table = pd.crosstab(df['pclass'], df['survived'])
print(contingency_table)


#perform chi-square test 
chi2, p ,dof, ex = chi2_contingency(contingency_table)

print('\n\nchi-squared statistics: ',chi2)
print('p-val', p)
print('degrees of freedom: ',dof)

print('\n\n')
print('expected values: \n',ex)

if p < 0.05:
    print('\nreject null hypothisis')
else:
    print('\nfail to reject null hypothesis')
```
