```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sns
```

```python
df = pd.read_csv('SP500 oil gold bitcoin.csv')
df
```

```python
corr = df[['BITCOIN','Gold','S&P500']].corr()
cov = df[['BITCOIN','Gold','S&P500']].cov()
```

```python
print(corr)
print(cov)
print(cov.to_string(float_format='{:.4f}'.format ))
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['BITCOIN'], df['Gold'], color='red',alpha=0.5, label='Bitcoin vs gold')
plt.scatter(df['BITCOIN'], df['S&P500'], color='green',alpha=0.5, label='Bitcoin S&P')
plt.scatter(df['S&P500'], df['Gold'], color='yellow',alpha=0.5, label='S&P vs Gold')

plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
# seaborn equivalent
sns.scatterplot(x= 'BITCOIN', y='Go' \
'ld', data=df, color = 'red', alpha = 0.5, label = 'Bitcoin vs Gold')

plt.figure(figsize=(6,4))
sns.heatmap(corr, annot=True, fmt=".2f", cmap='viridis')
plt.title('Overall correlation heatmap')
plt.xticks(rotation = 45)
plt.yticks(rotation = 0)
plt.show()

plt.figure(figsize=(6,4))
sns.heatmap(cov, annot=True, fmt=".2f", cmap='viridis')
plt.title('Overall cov heatmap')
plt.xticks(rotation = 45)
plt.yticks(rotation = 0)
plt.show()
```
