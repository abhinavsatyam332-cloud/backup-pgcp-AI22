```python
import pandas as pd
import numpy as np
from sklearn.model_selection import KFold , GridSearchCV
from sklearn.linear_model import ElasticNet
from sklearn.ensemble import RandomForestRegressor 
from xgboost import XGBRegressor
from lightgbm import LGBMRegressor
from sklearn.preprocessing import OneHotEncoder , StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer , make_column_transformer, make_column_selector
```

# Q1

```python
df = pd.read_csv(r"C:\Users\RAJESH\Downloads\Sacremento.csv")
df = df.drop('zip',axis=1)
df
```

```python
X ,y = df.drop('price',axis=1) , df['price']
```

```python
df.isna().sum()
```

```python
elnet = ElasticNet(random_state=26)
rf = RandomForestRegressor()
xbg= XGBRegressor()
lgbm = LGBMRegressor()

kfold = KFold()
```

```python
ohe = OneHotEncoder(drop='first',sparse_output=False, handle_unknown='ignore' ).set_output(transform='pandas')
scl = StandardScaler().set_output(transform='pandas')
ct = make_column_transformer(
                            (ohe,make_column_selector(dtype_include = object)),
                             ('passthrough',make_column_selector(dtype_include=['int64','float64'])),
                             verbose_feature_names_out=False).set_output(transform='pandas')
```

# ElasticNet

```python
pipe = Pipeline([('CT',ct),('SCL',scl),('ELNET',elnet)])
# pipe.get_params()
```

```python
params = {
    'ELNET__alpha': np.linspace(0.01,10,20),
    'ELNET__l1_ratio': np.linspace(0.1,1,10)
    
}
gcv = GridSearchCV(pipe, param_grid = params ,n_jobs=-1, cv = kfold , verbose = 0 ,scoring='r2')
gcv.fit(X,y)
```

```python
print(f'''
Best params: {gcv.best_params_}

Best Score: {gcv.best_score_}

''' )
```

# Random Forest

```python
pipe = Pipeline([('CT',ct),('RF',rf)])
# pipe.get_params()
```

```python
params = {
    'RF__max_features': [2,3,4,5,6,7,8,9],
    'RF__max_depth': [None, 2,3,4,5,6,7,8,9],
    'RF__n_estimators':[10,20,25,50,100]
}
gcv = GridSearchCV(pipe, param_grid = params ,n_jobs=-1, cv = kfold , verbose = 0 ,scoring='r2')
gcv.fit(X,y)
```

```python
print(f'''
Best params: {gcv.best_params_}

Best Score: {gcv.best_score_}

''' )
```

# XGB

```python
pipe = Pipeline([('CT',ct),('XGB',xbg)])
# pipe.get_params()
```

```python
params = {
    'XGB__n_estimators': [50, 100, 200],           
    'XGB__learning_rate': [0.01, 0.05, 0.1, 0.2],    
    'XGB__max_depth': [3, 4, 5, 6, 8],                  
    'XGB__gamma': [0, 0.1, 0.5]
}
gcv = GridSearchCV(pipe, param_grid = params ,n_jobs=-1, cv = kfold , verbose = 0 ,scoring='r2')
gcv.fit(X,y)
```

```python
print(f'''
Best params: {gcv.best_params_}

Best Score: {gcv.best_score_}

''' )
```

# LGBM

```python
pipe = Pipeline([('CT',ct),('LGM',lgbm)])
pipe.get_params()
```

```python
params = {
    'XGB__n_estimators': [50, 100, 200],           
    'XGB__learning_rate': [0.01, 0.05, 0.1, 0.2],    
    'XGB__max_depth': [3, 4, 5, 6, 8],             
    'XGB__subsample': [0.8, 1.0],                  
    'XGB__colsample_bytree': [0.8, 1.0],          
    'XGB__gamma': [0, 0.1, 0.5]
}
gcv = GridSearchCV(pipe, param_grid = params ,n_jobs=-1, cv = kfold , verbose = 0 ,scoring='r2')
gcv.fit(X,y)
```

```python
print(f'''
Best params: {gcv.best_params_}

Best Score: {gcv.best_score_}

''' )
```
