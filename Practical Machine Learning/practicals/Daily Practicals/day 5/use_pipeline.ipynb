
# pipeline

```python
from sklearn.pipeline import Pipeline
```

### 0. For Linear Regression

```python
df = pd.read_csv('ChemicalProcess.csv')
x = df.drop(columns = ['Yield'])
y = df['Yield']

X_train , X_test, y_train, y_test = train_test_split(x,y, test_size = 0.3 , random_state = 26)



imputer = SimpleImputer(strategy = 'mean').set_output(transform='pandas')
lr = LinearRegression()
pipe = Pipeline([('IMP', imputer), ('LR', lr)])               

pipe.fit(X_train,y_train)  # last tuple should be supervised ml if we want to call fit function 

y_pred = pipe.predict(X_test)
r2_score(y_test, y_pred)
```

### 1. For lasso

```python
df = pd.read_csv('ChemicalProcess.csv')
x = df.drop(columns = ['Yield'])
y = df['Yield']

X_train , X_test, y_train, y_test = train_test_split(x,y, test_size = 0.3 , random_state = 26)



imputer = SimpleImputer(strategy = 'mean').set_output(transform='pandas')
lasso = Lasso(alpha = 0.5, max_iter = 100000)
pipe = Pipeline([('IMP', imputer), ('LR', lasso)])               

pipe.fit(X_train,y_train)  # last tuple should be supervised ml if we want to call fit function 

y_pred = pipe.predict(X_test)
r2_score(y_test, y_pred)
```

### 3. For Elastic Net

```python
df = pd.read_csv('ChemicalProcess.csv')
x = df.drop(columns = ['Yield'])
y = df['Yield']

X_train , X_test, y_train, y_test = train_test_split(x,y, test_size = 0.3 , random_state = 26)

l1_ratios = np.linspace(0.001,1,20)
scores = []


imputer = SimpleImputer(strategy = 'mean').set_output(transform='pandas')

alphas = np.linspace(0.001, 10.0,20) 
l1_ratios = np.linspace(0.001,1,20)
for alp in alphas:
    for r in l1_ratios:

        elnet = ElasticNet(alpha=alp, l1_ratio = r)
        
        pipe = Pipeline([('IMP', imputer), ('ELNET', elnet)])   
        pipe.fit(X_train,y_train)

        y_pred = pipe.predict(X_test)
        scores.append([alp,r,r2_score(y_test, y_pred)])

 

# 2. Convert the list of results into a DataFrame
df_scores = pd.DataFrame(scores, columns=['alpha', 'l1_ratio', 'r2_score'])
df_scores.sort_values('r2_score',ascending = False,inplace = True)
df_scores.iloc[0]
```
