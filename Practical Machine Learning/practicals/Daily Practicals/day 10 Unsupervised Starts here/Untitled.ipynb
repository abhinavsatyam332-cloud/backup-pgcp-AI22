```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import linkage, dendrogram
from sklearn.cluster import AgglomerativeClustering
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
import os
os.chdir(r'E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets')
```

```python
df = pd.read_csv('milk.csv', index_col = 0)
df
```

```python
scaler = StandardScaler().set_output(transform='pandas')
df_scaled = scaler.fit_transform(df)
```

```python
hc = AgglomerativeClustering(n_clusters=3)
hc.fit(df_scaled)
# hc.labels_
silhouette_score(df_scaled,hc.labels_)
```

```python
clusters= [ 2,3,4,5,6,7]
scores = []
for c in clusters:
    hc = AgglomerativeClustering(n_clusters=c)
    hc.fit(df_scaled)
    # hc.labels_
    scores.append([c , silhouette_score(df_scaled,hc.labels_)])

df_score = pd.DataFrame(scores , columns = ['clusters', 'score'])
df_score.sort_values(by='score', ascending=False)
```
