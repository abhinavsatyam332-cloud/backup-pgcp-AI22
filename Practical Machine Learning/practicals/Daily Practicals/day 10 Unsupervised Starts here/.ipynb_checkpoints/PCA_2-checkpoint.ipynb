```python
import os
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

os.chdir(r"E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\Wisconsin")
```

```python
wisc = pd.read_csv('BreastCancer.csv',index_col=0)
wisc
```

```python
X , y = wisc.drop('Class', axis = 1), wisc['Class']
```

```python
scl = StandardScaler().set_output(transform='pandas')
scl_data = scl.fit_transform(X)

pca_comp= PCA().set_output(transform='pandas')
pca_data = pca_comp.fit_transform(scl_data)

pca_data
 
```

```python
pca_comp.explained_variance_ratio_*100
```

```python
np.cumsum(pca_comp.explained_variance_ratio_*100)
```

```python
pca2s = pca_data[['pca0','pca1']]
pca2s['Class'] = y
```

```python
plt.scatter(range(1,10) ,np.cumsum(pca_comp.explained_variance_ratio_*100) )
plt.plot(range(1,10) , np.cumsum(pca_comp.explained_variance_ratio_*100))
plt.xlabel('PCs')
plt.ylabel('Cumulative % age variation')
```

```python
sns.scatterplot(pca2s, x = 'pca0', y='pca1', hue='Class')
```

# SVM

```python
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import f1_score
```

```python
X_train , X_test , y_train ,y_test = train_test_split(X , y , test_size=0.3, stratify=y)
SVM = SVC(kernel='linear')
SVM.fit(X_train , y_train)
y_pred = SVM.predict(X_test)
f1_score(y_test, y_pred, pos_label = 'Malignant')
```
