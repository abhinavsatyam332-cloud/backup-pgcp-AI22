```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.tree import DecisionTreeClassifier , plot_tree, DecisionTreeRegressor
from sklearn.linear_model import LogisticRegression, LinearRegression, Lasso, Ridge
from sklearn.neighbors import KNeighborsClassifier , KNeighborsRegressor
from sklearn.ensemble import VotingClassifier,VotingRegressor

from sklearn.metrics import accuracy_score, r2_score
from sklearn.compose import ColumnTransformer , make_column_selector,make_column_transformer 
from sklearn.preprocessing import OneHotEncoder 
from sklearn.pipeline import Pipeline
from tqdm import tqdm
from sklearn.ensemble import BaggingClassifier , BaggingRegressor


from sklearn.model_selection import train_test_split
import warnings
warnings.filterwarnings("ignore")
import os 
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\Wisconsin')
```

```python
bcancer = pd.read_csv('BreastCancer.csv',index_col='Code')
bcancer
```

```python
X,y = bcancer.drop('Class', axis = 1) , bcancer['Class']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26, stratify=y)
```

# Bagging Classifier

```python
dtc = DecisionTreeClassifier(random_state=26)
lr = LogisticRegression()
knn = KNeighborsClassifier()

bagg = BaggingClassifier(estimator=lr , n_estimators=15 , random_state=26)

bagg.fit(X_train, y_train)
y_pred = bagg.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
bagg = BaggingClassifier(estimator=dtc , n_estimators=15 , random_state=26)

bagg.fit(X_train, y_train)
y_pred = bagg.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
bagg = BaggingClassifier(estimator=knn , n_estimators=15 , random_state=26)

bagg.fit(X_train, y_train)
y_pred = bagg.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
est = [dtc, lr, knn]
n_samples = [10,25,50]
scores = []
for e in est:
    for n in n_samples:
        bagg = BaggingClassifier(estimator=e , n_estimators=n , random_state=26)

        bagg.fit(X_train, y_train)
        y_pred = bagg.predict(X_test)
        scores.append([e, n , accuracy_score(y_test, y_pred)])

df_scores = pd.DataFrame(scores, columns=['Estimator','n','Score'])
df_scores.sort_values(by='Score',ascending=False)
```

```python
concrete_data = pd.read_csv('Concrete_Data.csv')
concrete_data
```

```python
X,y = concrete_data.drop('Strength', axis = 1) , concrete_data['Strength']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26)
```

```python

lr = LinearRegression()
lass = Lasso()
ridg = Ridge()
dtr = DecisionTreeRegressor(random_state=26)

bagg = BaggingRegressor(estimator=lr , n_estimators=15 , random_state=26)

bagg.fit(X_train, y_train)
y_pred = bagg.predict(X_test)
r2_score(y_test, y_pred)
```

```python
est = [lr,lass, ridg,dtr]
n_samples = [10,25,50]
scores = []
for e in est:
    for n in n_samples:
        bagg = BaggingRegressor(estimator=e , n_estimators=n , random_state=26)

        bagg.fit(X_train, y_train)
        y_pred = bagg.predict(X_test)
        scores.append([e, n , r2_score(y_test, y_pred)])

df_scores = pd.DataFrame(scores, columns=['Estimator','n','Score'])
df_scores.sort_values(by='Score',ascending=False)
```
