```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis , QuadraticDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer , make_column_selector,make_column_transformer
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder , StandardScaler
from sklearn.pipeline import Pipeline, make_pipeline
from sklearn.svm import SVC
from tqdm import tqdm
import os
import warnings
warnings.filterwarnings('ignore')
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\Vehicle Silhouettes')
```

```python
df = pd.read_csv('Vehicle.csv')
df
```

```python
X , y = df.drop('Class', axis=1), df['Class']
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26 ,stratify=y)
```

```python
cs = np.linspace(0.001, 10,20)
scaler = StandardScaler().set_output(transform='pandas')
scores  =[]
dfs=['ovo','ovr']

for c in tqdm(cs):
    for f in dfs:
        svm = SVC(kernel='linear',C = c,decision_function_shape=f)
        pip = Pipeline([('scaler',scaler),('svm',svm)])
        pip.fit(X_train, y_train)
        y_pred = pip.predict(X_test)
        scores.append([ c,f, accuracy_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores , columns=['c','Decision','score'])

df_scores.sort_values(by='score',ascending=False)
```

```python
cs = np.linspace(0.001, 10,20)
gs = np.linspace(0.1, 5,10)
scaler = StandardScaler().set_output(transform='pandas')
scores  =[]
dfs=['ovo','ovr']

for c in tqdm(cs):
    for f in dfs:
        for g in gs:
            svm = SVC(kernel='rbf',C = c,decision_function_shape=f,gamma=g)
            pip = Pipeline([('SCL',scaler),('svm',svm)])
            pip.fit(X_train, y_train)
            y_pred = pip.predict(X_test)
            scores.append([ c,f,g, accuracy_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores , columns=['c','Decision','Gamma','score'])

df_scores.sort_values(by='score',ascending=False)
```
