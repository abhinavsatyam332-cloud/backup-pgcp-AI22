# Model Ensembling On BreastCancer Dataset
apply DecisionTreeClassifier  , LogisticRegression ,KNeighborsClassifier
calculate indivisual probability for each model    
use hard , soft voting


```python
breast_cancer = pd.read_csv('BreastCancer.csv',index_col='Code')
breast_cancer
```

```python
X,y = breast_cancer.drop('Class', axis = 1) , breast_cancer['Class']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26,stratify=y)
```

```python
dtc = DecisionTreeClassifier(random_state=26)
lr = LogisticRegression()
knn = KNeighborsClassifier()

voting = VotingClassifier([('TREE', dtc),('LR',lr),('KNN',knn)])
voting.fit(X_train,y_train)
y_pred = voting.predict(X_test)
accuracy_score(y_test,y_pred)
```

# Find best parameters   

default voting = 'hard'

```python
depths = [None, 2,3,4,5,6,7]
solvers = ['lbfgs','saga','sag','liblinear']
ks = [2,5,10]
scores = []

for d in depths:
    for s in solvers:
        for k in ks:
            dtc = DecisionTreeClassifier(random_state=26, max_depth=d)
            lr = LogisticRegression(solver=s)
            knn = KNeighborsClassifier(n_neighbors=k)

            voting = VotingClassifier([('TREE', dtc),('LR',lr),('KNN',knn)])
            voting.fit(X_train,y_train)
            y_pred = voting.predict(X_test)

            scores.append([d,s,k,accuracy_score(y_test,y_pred)])

df_scores = pd.DataFrame(scores, columns=['Depth','Solver','K','Score'])
df_scores.sort_values(by='Score',ascending=False)
```

```python
depths = [None, 2,3,4,5,6,7]
solvers = ['lbfgs','saga','sag','liblinear']
ks = [2,5,10]
scores = []

for d in depths:
    for s in solvers:
        for k in ks:
            dtc = DecisionTreeClassifier(random_state=26, max_depth=d)
            lr = LogisticRegression(solver=s)
            knn = KNeighborsClassifier(n_neighbors=k)

            voting = VotingClassifier([('TREE', dtc),('LR',lr),('KNN',knn)], voting='soft', weights=[0.3,1,1])   # soft use .predict_proba for each model then takes avg
            voting.fit(X_train,y_train)
            y_pred = voting.predict(X_test)

            scores.append([d,s,k,accuracy_score(y_test,y_pred)])

df_scores = pd.DataFrame(scores, columns=['Depth','Solver','K','Score'])
df_scores.sort_values(by='Score',ascending=False)
```
