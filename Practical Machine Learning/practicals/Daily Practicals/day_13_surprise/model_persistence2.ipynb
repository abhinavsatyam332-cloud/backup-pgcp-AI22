# load pretrained model and use it
Model Persistence (pickle.load): Loading a previously trained and saved model object (svm_model.pkl) into memory so you don't have to retrain it from scratch.


Model Inference (loaded_model.predict(df)): Feeding unseen test data (tst_cancer.csv) into the trained model to generate outputs (Malignant or Benign).



Training / Fitting (model.fit(...)): The model learns the underlying patterns, weights, and decision boundaries from your historical dataset.

Inference (model.predict(...)): The model applies those already-learned patterns to new, unseen data to generate predictions or classifications.
