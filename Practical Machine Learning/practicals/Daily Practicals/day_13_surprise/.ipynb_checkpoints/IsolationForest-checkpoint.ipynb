```python
import pandas as pd
import numpy as np
import os
from sklearn.ensemble import IsolationForest
os.chdir('E:\PGCP-AI\PML\Datasets')
```

```python
df = pd.read_csv('milk.csv', index_col = 0)
df
```

```python
df = pd.read_csv('nutrient.csv', index_col = 0)
df
```

```python


clf = IsolationForest(contamination = 0.05 , random_state=26)
clf.fit(df)

predictions = clf.predict(df)
print(predictions)
print('% age of outliers = '+str((predictions<0).mean()*100)+ ' %')
abn_ind = np.where(predictions < 0)
 
abn_ind
```
