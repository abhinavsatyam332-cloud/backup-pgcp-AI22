Model Serialization

Serialization (pickle.dump): Converting an in-memory Python object—such as your trained LinearRegression model, 
                             weights, intercept, and internal attributes—into a byte stream so it can be saved as a binary file (.pkl) on your                                hard drive.

Deserialization (pickle.load): The reverse process of reading the binary file back into memory to restore the exact Python object later.
    

```python
import os
import pandas as pd
import pickle
from sklearn.linear_model import LinearRegression
os.chdir('E:\PGCP-AI\PML\Datasets')
```

```python
insurance = pd.read_csv('Insure_auto.csv', index_col = 0)
X , y = insurance.drop('Operating_Cost' , axis = 1), insurance['Operating_Cost']
lr = LinearRegression()
lr.fit(X , y)
pkfile = open('E:\PGCP-AI\PML\insurance_lr.pkl', 'wb')   # creates file reference (it is projects artifact)
pickle.dump(lr,pkfile)
pkfile.close()
```

```python
pkfile = open(r"E:\PGCP-AI\PML\insurance_lr.pkl", "rb")
loaded_model = pickle.load(pkfile)

unlab = pd.DataFrame({'Home':[500], 'Automobile':[1000]})
loaded_model.predict(unlab)
```
