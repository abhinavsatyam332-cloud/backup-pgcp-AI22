```python
# embedding on content of news
import pandas as pd
import numpy as np
from tqdm import tqdm
from sentence_transformers import SentenceTransformer
```

```python
df = pd.read_csv('historic_articles.csv')
df
```

```python
df.shape
df.columns
```

```python
model = SentenceTransformer('BAAI/bge-small-en-v1.5')
```

```python
df_indian_movies = df[(df['category'].str.contains('india')) ]
df_indian_movies
```

```python
all_content = list(df_indian_movies['content'].fillna(''))

out_embs = []  # this are vector embeddings which are generally stored in vector Databases like pinecone , Qdrant

for i in tqdm(all_content):
  enmd = model.encode(i)
  out_embs.append(enmd)
out_embs = np.array(out_embs)
```

```python

df_embd = pd.DataFrame(out_embs )
df_embd
```

```python
# see we have created df_indian_movies   and created df_embd   for that df_indian_movies
# model.encode always returns 384 columns as we are using SentenceTransformer('BAAI/bge-small-en-v1.5')
```

```python
from sklearn.cluster import HDBSCAN
from sklearn.metrics import silhouette_score
```

```python
hdbscan = HDBSCAN(min_cluster_size=2 )
hdbscan.fit(out_embs)
df_embd['hdbscan_cluster'] = hdbscan.labels_
df_embd['hdbscan_cluster'].value_counts()
```
