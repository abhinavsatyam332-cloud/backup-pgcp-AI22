```python
import pandas as pd
from mlxtend.frequent_patterns import apriori
from mlxtend.frequent_patterns import association_rules



import os 
os.chdir(r'E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets')

import warnings
warnings.filterwarnings("ignore")
```

```python
'''
Binary Format: Association rule mining only cares about whether an item was present (True) or absent (False), not how many units were bought.

Algorithm Compatibility: The function relies on boolean matrix operations to quickly calculate support and count itemsets.
'''
```

```python
fp_df = pd.read_csv('Faceplate.csv',index_col=0)
fp_df = fp_df.astype(bool)
fp_df
```

```python
itemsets = apriori(fp_df, min_support=0.2, use_colnames=True)
'''
This means the algorithm only keeps itemsets (like individual items or combinations)
that appear in at least 20% of the total transactions in your dataset.
'''
itemsets
```

```python
rules = association_rules(itemsets , metric='confidence' , min_threshold=0.5)
rules.sort_values(by=['confidence','lift','support'], ascending=False)
```

```python
relv_rules= rules[rules['lift']>1]
relv_rules.sort_values(by=['confidence','lift','support'], ascending=False)
```
