```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.naive_bayes import BernoulliNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score

import os
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\Telecom')
```

```python
telecom = pd.read_csv('Telecom.csv')
X, y = telecom.drop('Response', axis = 1), telecom['Response']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26, stratify=y)

ohc = OneHotEncoder(sparse_output = False, drop = 'first').set_output(transform = 'pandas')
nb = BernoulliNB()

pipe = Pipeline([('OHE', ohc),('NB', nb)])
pipe.fit(X_train, y_train)

y_pred = pipe.predict(X_test)
accuracy_score(y_test, y_pred)
```

```python
# Get predicted probabilities for each class on the test set
pipe.predict_proba(X_test)       # predict prob is in all where classification algo
```
