```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder
from sklearn.naive_bayes import BernoulliNB,GaussianNB
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
from sklearn.impute import SimpleImputer 
from sklearn.compose import make_column_transformer,make_column_selector
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
import os
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\student_performance')
```

```python
studper = pd.read_csv('student_dataset_10000_rows.csv')
studper
```

```python
X,y = studper.drop('placement_status', axis = 1) , studper['placement_status']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26,stratify=y)
```

```python
nb = GaussianNB()
 
nb.fit(X_train, y_train)
 
y_pred = nb.predict(X_test) # gen outcome with highest prob
accuracy_score(y_test, y_pred)
```

```python
nbs = range(1,15)
res =[] 
for i in nbs:
    
    knn = KNeighborsClassifier(n_neighbors = i)
    knn.fit(X_train, y_train)
 
    y_pred = knn.predict(X_test) # gen outcome with highest prob
    res.append([i , accuracy_score(y_test, y_pred)])

rs = pd.DataFrame(res , columns=['K', 'accuracy'])
rs.sort_values(by='accuracy', ascending = False)
    
```

### After scaling check the accuracy

```python
scaler = StandardScaler().set_output(transform = 'pandas')
scaled_x_train  = scaler.fit_transform(X_train)
scaled_x_test  = scaler.transform(X_test)


nb = GaussianNB()
 
nb.fit(scaled_x_train, y_train)
 
y_pred = nb.predict(scaled_x_test) # gen outcome with highest prob
accuracy_score(y_test, y_pred)
```

```python
nbs = range(1,15)
res =[] 
for i in nbs:
    
    knn = KNeighborsClassifier(n_neighbors = i)
    knn.fit(scaled_x_train, y_train)
 
    y_pred = knn.predict(scaled_x_test) # gen outcome with highest prob
    res.append([i , accuracy_score(y_test, y_pred)])

rs = pd.DataFrame(res , columns=['K', 'accuracy'])
rs.sort_values(by='accuracy', ascending = False)
    
```
