```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis , QuadraticDiscriminantAnalysis
from sklearn.linear_model import LogisticRegression
import matplotlib.pyplot as plt
from sklearn.compose import ColumnTransformer , make_column_selector,make_column_transformer
import seaborn as sns
from sklearn.preprocessing import OneHotEncoder , StandardScaler
from sklearn.pipeline import Pipeline, make_pipeline
import os
import warnings
warnings.filterwarnings('ignore')
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets\Cases\Image_Segmentation')
```

```python
imgseg = pd.read_csv('Image_Segmentation.csv')
imgseg
```

```python
X, y = imgseg.drop('Class', axis = 1), imgseg['Class']

X_train, X_test, y_train, y_test = train_test_split(X,y, test_size = 0.3, random_state=26, stratify=y)
 
```

```python
solvers=['lbfgs','newton-cg','newton-cholesky','sag','saga']
# liblinear   - not applied
scores  =[]


for s in solvers:
    lr  = LogisticRegression(solver=s)
    lr.fit(X_train, y_train)
    y_pred = lr.predict(X_test)
    scores.append([s , accuracy_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores , columns=['solver','score'])
df_scores
```

```python
solvers=['lbfgs','newton-cg','newton-cholesky','sag','saga']
# liblinear   - not applied
cs = np.linspace(0.001, 10,20)
scores  =[]


for s in solvers:
    for c in cs:
        lr  = LogisticRegression(solver=s,C = c)
        lr.fit(X_train, y_train)
        y_pred = lr.predict(X_test)
        scores.append([s , c, accuracy_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores , columns=['solver','c','score'])
df_scores.sort_values(by='score',ascending=False)
```

# When l1_ratio = 1 

```python
 
# liblinear   - not applied
cs = np.linspace(0.001, 10,20)
scores  =[]

 
for c in cs:
    lr  = LogisticRegression(solver='saga',C = c,l1_ratio=1)
    lr.fit(X_train, y_train)
    y_pred = lr.predict(X_test)
    scores.append([ c, accuracy_score(y_test, y_pred)])
df_scores = pd.DataFrame(scores , columns=['c','score'])
df_scores.sort_values(by='score',ascending=False)
```
