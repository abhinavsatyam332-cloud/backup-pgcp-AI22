```python
# verbose_features_names_out removes the prefix like which encoder used, passthrough


df = pd.read_csv('Exp_salaries.csv')
df

x = df.drop('Salary', axis=1)
y = df['Salary']

ohc = OneHotEncoder(sparse_output = False, drop = 'first').set_output(transform = 'pandas')

ct = make_column_transformer(
                            (ohc,           make_column_selector(dtype_include = object)),
                            ('passthrough', make_column_selector(dtype_include =['int64','float64'])),
                             verbose_feature_names_out = False)
ct = ct.set_output(transform = 'pandas')
x_ohe = ct.fit_transform(x)
x_ohe
```

```python
df = pd.read_csv('Exp_Salaries.csv')
X, y = df[['Gender']], df['Salary']
ohe = OneHotEncoder(sparse_output=False).set_output(transform='pandas')        # drop='first'   automatically drops the first category column for every categorical feature it encodes.
x_ohe = ohe.fit_transform(X)
x_ohe = x_ohe.drop(columns=['Gender_Male'])

lr = LinearRegression()
lr.fit(x_ohe,y)
lr.coef_ , lr.intercept_ 
 
```
