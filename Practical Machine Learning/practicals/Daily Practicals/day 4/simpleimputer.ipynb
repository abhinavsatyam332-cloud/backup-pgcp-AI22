```python
import numpy as np 
import pandas as pd
from sklearn.impute import SimpleImputer

import os
os.chdir(r'E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets')
```

```python
job = pd.read_csv('JobSalary2.csv', index_col = 0)
job
```

```python
job.mean()
```

### Mean Imputation

```python
mu = dict(job.mean())
job.fillna(mu)
```

### Median Imputation

```python
mu = dict(job.median())
job.fillna(mu)
```

```python
# SImpleImputer
```

```python
imputer  = SimpleImputer(strategy='mean').set_output(transform='pandas')
imputer.fit(job) # calculates the statistics
print(imputer.statistics_)

imputer.transform(job) # fills the na with stat values
```
