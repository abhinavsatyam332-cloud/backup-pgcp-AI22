```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

import os
os.chdir(r'E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets')
```

## On milk dataset

```python
df = pd.read_csv("monthly-milk-production-pounds-p.csv",index_col =  0)
df.index = pd.to_datetime(df.index).to_period('M')
# df.index.dtype
df.plot()
plt.title('Monthly milk production')
plt.xlabel('Months')
plt.ylabel('Milk Production in pounds')

# resample() groups time-series data into wider time blocks (like monthly to quarterly).
# think of it like grouby then aggregating
downsampled = df.resample('Q').sum()

downsampled.index.rename('Quarter',inplace=True)
downsampled.head()

downsampled.plot(marker='o')
plt.grid()
plt.title('Quarterly Milk Production in pounds')
plt.ylabel('Milk Production in pounds')
```

## On Electric_Production dataset

```python
electric = pd.read_csv('Electric_Production.csv',index_col =0 )
electric
```

```python
 
electric.index = pd.to_datetime(electric.index).to_period('M')
electric
```

```python
electric.plot()
plt.title('Monthly electric production')
plt.xlabel('Months')
plt.ylabel('electric Production ')
```

```python
downsampled = electric.resample('Q').sum()

downsampled.index.rename('Quarter',inplace=True)
downsampled.head()

downsampled.plot()
plt.title('Electric production')
plt.xlabel('Months')
plt.ylabel('elec Production in pounds')
```

## Logarithmic Graph

```python
electric['IPG2211A2N'] = np.log(electric['IPG2211A2N'])
electric.plot()
plt.title('Log scaled electricity production')
plt.xlabel('Months')
plt.show()
```
