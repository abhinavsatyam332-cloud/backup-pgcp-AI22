```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt

from statsmodels.tsa.seasonal import seasonal_decompose
import os
os.chdir(r'E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets')
```

# seasonal decompose on milk-production

```python
df = pd.read_csv('monthly-milk-production-pounds-p.csv' )
df.plot()
plt.show()
df
```

```python
series = df['Milk']
result = seasonal_decompose(series, model='additive', period=12)     # model = additive multiplicative
result.plot()
plt.title('Additive Decomposition')
plt.show()
```

```python
series = df['Milk']
result = seasonal_decompose(series, model='multiplicative', period=12)
result.plot()
plt.title('multiplicative Decomposition')
plt.show()
```
