```python
import os
import pandas as pd
import numpy as np 
from xgboost import XGBRegressor  

from sklearn.model_selection import KFold ,cross_val_score, cross_val_predict
from tqdm import tqdm 
from sklearn.model_selection import GridSearchCV
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import make_column_selector , make_column_transformer

import warnings
warnings.filterwarnings("ignore")

os.chdir(r"C:\Users\RAJESH\Desktop\ss\Datasets\Cases\Concrete_Strength")
```

```python
concrete_data = pd.read_csv("Concrete_Data.csv")
X , y = concrete_data.drop('Strength' , axis = 1) , concrete_data['Strength']
```

```python
xgb = XGBRegressor(random_state=26)
kfold = KFold(random_state=26, n_splits=5, shuffle=True)
params = {
    "learning_rate":np.linspace(0.01,0.9,10),
    "max_depth":[None,2,3,4],
    "n_estimators":[25,50,100]
}

gcv = GridSearchCV(xgb,param_grid = params , cv = kfold, scoring='r2')

gcv.fit(X, y)
print(f'Best Params : {gcv.best_params_}')
print(f'Best Score : {gcv.best_score_}')
```

```python
concrete_data_tst = pd.read_csv("testConcrete.csv")
concrete_data_tst
```

```python
xgb = XGBRegressor(**gcv.best_params_, random_state=26)
xgb.fit(X,y)

xgb.predict(concrete_data_tst)
```
