```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import root_mean_squared_error    # use tab to get auto suggestion 
from sklearn.preprocessing import PolynomialFeatures
import os
os.chdir('E:\Backup Files Abhinav_Satyam\Practical Machine Learning\Sanjay Sir Git Repo\Datasets')
```

```python
df = pd.read_csv('pizza.csv')
x = df[['Promote']]
y = df['Sales']
#     2.300e+01   ->  2.3 
poly = PolynomialFeatures(degree = 3,include_bias = False).set_output(transform = 'pandas')

x_poly =poly.fit_transform(x)
x_poly


```

```python
model = LinearRegression()
model.fit(x_poly,y)

print('Coef: ', model.coef_)
print('Intercept: ', model.intercept_)
```

```python
df = pd.read_csv('Boston.csv')
X , y = df.drop('medv',axis = 1), df['medv']


poly = PolynomialFeatures(degree = 2,include_bias = False).set_output(transform = 'pandas')
x_poly =poly.fit_transform(X)
x_train , X_test, y_train, y_test = train_test_split(x_poly, y, random_state = 26, test_size = 0.3)

lr = LinearRegression()
lr.fit(x_train, y_train)
y_pred = lr.predict(X_test)

root_mean_squared_error(y_test, y_pred)
```
