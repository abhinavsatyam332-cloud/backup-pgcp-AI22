1.	Draw line chart to show urban and rural population of India over the years using population.csv [rural 	population is the difference between total population and urban population]

```python
import pandas as pd
import numpy as np
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv('Population.csv')
df
```

```python
df['Rural Population'] = df['Population'] - df['Urban Population']
```

```python
df['Year'].max()
```

```python

plt.plot(df['Year'],df['Urban Population'],ls = '--' ,lw = 1)
plt.plot(df['Year'],df['Rural Population'],ls = '--' ,lw = 1)
plt.ylabel('Population')
plt.xlabel('year')
plt.legend(['Urban Population','Rural Population'])
plt.figure(figsize=(2,2 ))
plt.show()

# plots urban population vs rural population over year using dashed lines
```

2. 	Using a Bar chart display India's total, urban and rural population from year 2020 to 2025

```python
df['Year'].loc[13:] 
```

```python
filtered_data  = df[ (df['Year'] > 2019) & (df['Year'] < 2026 ) ]
filtered_data
# filter out the dataframe for years between 2019 and 2026
```

```python
xwidth = 0.3
x_year = filtered_data['Year']
plt.figure(figsize=(12,6))
b1 = plt.bar(x_year-xwidth, filtered_data['Rural Population'] ,width =xwidth, label='Rural Population' ,color="#adc527")
b2 = plt.bar(x_year, filtered_data['Urban Population'] ,width =xwidth, label='Urban Population',color="#143686")
b3 = plt.bar(x_year+xwidth, filtered_data['Population'],width =xwidth, label='Total Population',color="#D9CEEBD7")

plt.bar_label(b1, fmt='{:,.0f}', padding=-70, rotation=85, fontsize=7)
plt.bar_label(b2, fmt='{:,.0f}', padding=-70, rotation=85, fontsize=7)
plt.bar_label(b3, fmt='{:,.0f}', padding=-70, rotation=85, fontsize=7)



plt.xticks(rotation=30)
plt.xlabel("population")
plt.ylabel("total")
plt.title("Solar Energy Installed Capacity by State")
plt.legend()


plt.show()

# this graph shows Rural, Urban, and Total Population side-by-side.
# stored x data means df[year] series of filtered data
# set size of figure for plotting
# created group bar chart  using  Rural Population , Urban Population and total Population
# give the labels to each sub bars with their their values
# then rotated xticks with 30  degree angle
# set xlable, ylable , title, legend, and last show method
```

```python
xwidth = 0.3
x_year = df['Year'].loc[13:]

plt.bar(x_year, df['Population'].loc[13:],width =0.6, label='Total Population',color="#D9CEEBD7")
plt.bar(x_year-xwidth/2, df['Rural Population'].loc[13:],width =xwidth, label='Rural Population' ,color="#adc527")
plt.bar(x_year+xwidth/2, df['Urban Population'].loc[13:],width =xwidth, label='Urban Population',color="#143686")


plt.xticks(rotation=30)
plt.xlabel("population")
plt.ylabel("total")
plt.title("Solar Energy Installed Capacity by State")
plt.legend()

plt.show()
```

3.	Using a pie chart show distribution of top six selling cars [Create data as required]

```python
import random

random.seed(42)
car_names = [
    'Nexon', 'WagonR', 'Swift', 'Baleno', 'Brezza',
    'Punch', 'Creta', 'Venue', 'i20', 'Seltos',
    'Sonet', 'Thar', 'Scorpio', 'XUV700', 'Alto',
    'Ertiga', 'Fronx', 'Dzire', 'Harrier', 'Tiago'
]
data = {
    'name':  car_names,
    'price': [random.randint(15000, 50000) for _ in car_names],
    'sell count': [random.randint(100, 1000) for _ in car_names]
}

df = pd.DataFrame(data)
df
```

```python
carsdf = df.sort_values('sell count',ascending=False).head(6)
carsdf
# sorting dataframe based on sell count column in descending order
```

```python
plt.pie(carsdf['sell count'],
        labels=carsdf['name'],
        colors=['red','orange','blue','brown','violet','green'],
        shadow= 1,
        autopct='%1.2f%%',
        explode=[0.2,0,0,0,0,0],
        wedgeprops={'edgecolor':'black','linewidth':0.8, 'width':1.0,'linestyle':'--'},

            )
plt.title('Sell count of cars')
plt.legend()
plt.tight_layout()
plt.show()

# Creates a pie chart showing market share for top 6 selling car models.
# Highlights the top seller with explode, custom colors, percentage labels, and dashed edges.
```
