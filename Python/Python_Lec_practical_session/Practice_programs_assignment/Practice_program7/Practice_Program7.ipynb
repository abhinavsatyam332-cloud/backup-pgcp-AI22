```python
import plotly.express as px

# Load Gapminder dataset
df = px.data.gapminder()

# 1. Filter for European countries and calculate Total GDP
df_europe = df[df['continent'] == 'Europe'].copy()
df_europe['gdp'] = df_europe['gdpPercap'] * df_europe['pop']

# 2. Horizontal Bar Chart for 2007 (Percentage of Total European GDP)
df_2007 = df_europe[df_europe['year'] == 2007].copy()
df_2007['gdp_pct'] = (df_2007['gdp'] / df_2007['gdp'].sum()) * 100
df_2007 = df_2007.sort_values('gdp_pct', ascending=True)

fig_bar = px.bar(
    df_2007,
    x='gdp_pct',
    y='country',
    orientation='h',  # Horizontal orientation for clean country labels
    text=df_2007['gdp_pct'].apply(lambda x: f"{x:.1f}%"),  # Custom percentage text
    color='gdp_pct',
    color_continuous_scale='Blues',
    title='European Countries GDP Share (%) in 2007',
    labels={'gdp_pct': 'GDP Percentage (%)', 'country': 'Country'}
)
fig_bar.update_traces(textposition='outside')
fig_bar.show()

# 3. Stacked Bar Chart with Bar Norm (% Share across years)
fig_norm = px.bar(
    df_europe,
    x='year',
    y='gdp',
    color='country',
    barmode='stack',
    title='European Countries GDP Share (%) Over Time (Normalized Stacked)',
    labels={'gdp': 'GDP Share (%)', 'year': 'Year', 'country': 'Country'}
)
fig_norm.update_layout(barnorm='percent')  # Normalizes stack height to 100%
fig_norm.show()
```
