```python
df = sns.load_dataset('penguins')
df.head()
```

1. 

```python
df[
    (df['species']=='Chinstrap') & 
    (
        (df['flipper_length_mm'] >=180) & (df['flipper_length_mm'] <190)
    )
].count()
```

```python
sns.histplot(data = df,
            x='flipper_length_mm',
            # bins=5,
            hue = 'species',
            multiple='stack',
            bins = [170, 180, 195, 200, 210, 220, 230], 
            )
plt.grid(True)
```

```python
sns.histplot(data = df,
            x='flipper_length_mm',
            bins=5,
            hue = 'species',
            element = 'step')
```

2. Draw a bivariant KDE plot to show relationship between horsepower and displacement [use mpg dataset] – Write down your 	interpretations 

```python
df = sns.load_dataset('mpg')
df.head()
```

```python
sns.kdeplot(df,x='displacement', y='horsepower',fill=True,cmap='Blues',cbar=True)
```

3.	Draw a correlation heatmap for diamonds dataset, add annotation, change colour pallet and write your interpretation 

```python
df = sns.load_dataset('diamonds')
df
```

```python
df = df.loc[ : ,'depth':'z']
df
```

```python
sns.heatmap(df.corr(),cmap='coolwarm')
```
