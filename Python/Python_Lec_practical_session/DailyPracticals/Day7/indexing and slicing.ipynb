```python
arr = np.array([
    # Box 1 (Now has 3 rows and 4 columns)
    [[1,  2,  9,  0],
     [3,  4, 10,  1],
     [5,  6, 11,  2]],

    # Box 2 (Must match the same dimensions: 3 rows and 4 columns)
    [[7,  8, 12,  3],
     [9,  10, 13, 4],
     [11, 12, 14, 5]]
])

print(arr.shape)  # Output will be: (2, 3, 4)
```

```python
arr[[0,1],[0,1]]
```

```python
arr[::2, ::2]
```

```python
arr.shape
```

```python
arr[::, 0:2, 1:3]
```

```python
i_matrix = [0,0,0,0,1,1,1,1]
i_rows = [0, 0, 2, 2, 0, 0, 2, 2]
i_cols = [0, 2, 0, 2, 0, 2, 0, 2]
arr[i_matrix, i_rows, i_cols]
```
