2.1 normalize a NumPy array [ to normalize means scaling the values of an array into a specific range, often [0, 1]]

```python
arr = np.random.random(20)*50
print(arr)
```

```python
normalized = (arr - arr.min()) / (arr.max()-arr.min())
print(normalized)
```

2.2 Create a 5×5 array with 1 on border and 0 inside

```python
arr = np.zeros((6,3),dtype=int)
arr[0] = 1
```

```python
arr[0] = 1
arr[:,0] = 1
arr[:,len(arr[0])-1] = 1
arr[len(arr)-1] = 1
print(arr)
```

2.4 Broadcast a 1D array over a 2D array [all possible scenarios]

```python
arr1 = np.random.randint(0,10,16).reshape(4,4)
print(arr1)
```

```python
arr2 = np.random.randint(0,5,4)
print(arr2)
```

```python
arr1 + arr2
```

```python

result_rows = arr1 + arr2[:, np.newaxis]
print("arr1 = ",arr1)
print("arr2 = ",arr2)

print("Scenario 2 (Across Columns):\n", result_rows)
```
