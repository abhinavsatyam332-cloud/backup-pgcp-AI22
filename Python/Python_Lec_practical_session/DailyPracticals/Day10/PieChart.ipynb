```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
count = [35, 25, 20, 12, 8]
langs = ["Python", "JavaScript", "Java", "C++", "Ruby"]
```

```python
plt.pie(count     )
```

```python
plt.pie(count,labels=langs,radius=1,shadow=10,autopct='%1.1f%%' )
```

```python
plt.pie(count,
        labels=langs,
        colors=['orange','green','pink','blue','yellow'],
        radius=1,
        shadow=10,
        autopct='%1.1f%%' ,
        textprops={'fontsize':12,'color':'brown'}, 
        explode=[0,0,0,0.3,0],
        pctdistance=0.50,
        wedgeprops={'edgecolor':'black','linewidth':0.8, 'width':0.9,'linestyle':'--'},
        labeldistance=1.2,
        counterclock = True,
        normalize=True
        )
plt.show()
```
