```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("survey.csv")
```

```python
ages_x = list(range(18, 30))
dev_y = [42, 55, 68, 74, 81, 95, 110, 125, 138, 150, 162, 175]
py_dev_y = [18, 27, 35, 44, 52, 63, 76, 88, 101, 115, 129, 142]
```

```python
plt.bar(ages_x, dev_y)
```

```python
x_vals = np.arange(len(ages_x))
print(x_vals)  
b_width = 0.4
```

```python
plt.bar(x_vals - b_width/2, dev_y,  color = "#6943C4", width = b_width, label = "Developer salary")
plt.bar(x_vals + b_width/2, py_dev_y,  color = "#39a092", width = b_width, label = "Python Developer salary")
plt.xticks(ticks = x_vals, labels = ages_x)
plt.xlabel("Age")
plt.ylabel("Salary in $")
plt.title("Developer Salaries by Age")
plt.legend()
plt.show()
```
