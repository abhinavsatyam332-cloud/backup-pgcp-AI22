```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= sns.load_dataset('iris')
```

```python
sns.get_dataset_names()
```

```python
df
```

```python
sns.histplot(df,x = 'petal_length')
```

```python
print("\n\nFor setosa  ",df.groupby('species').get_group('setosa')['petal_length'].head(2))
print("\n\nfor versicolor ",df.groupby('species').get_group('versicolor')['petal_length'].head(2))
print("\n\nfor virginica  ",df.groupby('species').get_group('virginica')['petal_length'].head(2))

```

```python
sns.histplot(df,x='petal_length',hue='species')
```

```python
sns.histplot(df,
             x = 'petal_length',
             hue='species',
             multiple='stack')
```

```python
sns.histplot(df,
             x = 'petal_length',
             hue='species',element='step')
```

```python
sns.displot(df,
             x = 'petal_length',
             hue='species',
            col ='species')
```
