```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('mpg')
```

```python
df = df.loc[:,:'model_year']
```

```python
df.corr()
```

```python
sns.heatmap(df.corr())
```

```python
print(sns.heatmap(df.corr() , cmap='crest')) 
```

```python
sns.heatmap(df.corr(),cmap='RdBu', annot=True,fmt=".1f",linewidths=0.5)
```

```python
sns.heatmap(df.corr(), annot=True,fmt=".1f",linewidths=0.5)
```

```python
sns.heatmap(df.corr(), annot=True,fmt=".1f",linewidths=0.5,cmap='coolwarm')  # Blues Greens Purples
```

```python
sequential_cmaps = [
    'crest', 'flare', 'mako', 'rocket', 'viridis', 'magma', 'inferno',
    'Blues', 'Greens', 'Purples', 'Reds', 'Greys', 'Oranges'
]

# Diverging Palettes (Great for correlations)
diverging_cmaps = [
    'coolwarm', 'vlag', 'icefire', 'RdBu', 'BrBG'
]

for colour_map_nm in diverging_cmaps:
    
    print(sns.heatmap(df.corr(), annot=True,fmt=".1f",linewidths=0.5,cmap=colour_map_nm ))
    plt.title('cmap = '+colour_map_nm)  
    plt.show()
    plt.tight_layout()
```
