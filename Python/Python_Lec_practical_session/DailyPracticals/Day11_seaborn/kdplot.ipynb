```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df= sns.load_dataset('tips')
df
```

```python
sns.kdeplot(df, x = 'total_bill')
```

```python
sns.kdeplot(df, x = 'total_bill' , bw_adjust= 0.20 )          # bandwidth
```

```python
sns.kdeplot(df, x = 'total_bill', hue='sex')
```

```python
sns.kdeplot(df, x = 'total_bill', hue='sex',fill=True)
```

```python
sns.kdeplot(df, x = 'total_bill', hue='time',fill=True)
```

```python
sns.kdeplot(df, x = 'total_bill', hue='sex',fill=True, multiple='stack')
```
