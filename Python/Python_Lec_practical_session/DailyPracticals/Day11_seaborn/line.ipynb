```python
sns.lineplot(df,x = 'year', y = 'passengers',ci=100)         # default confidence interval is 95  
'''
A Confidence Interval (CI) is a shaded band on a chart that shows how certain or precise an estimate is

The Line: Represents the average (mean) of your data points.
The Shaded Band (CI): Shows the range where the true average likely falls.
The Width: A narrow band means your data is consistent and you can be highly confident in the line. A wide band means your data varies a lot, making the line less certain.

'''
```

```python
sns.lineplot(df,x = 'year', y = 'passengers',hue='month',ci=80)         # default confidence interval is 95  
```

```python
sns.lineplot(df,x = 'year', y = 'passengers',size='month',hue='month',ci=80)
```
