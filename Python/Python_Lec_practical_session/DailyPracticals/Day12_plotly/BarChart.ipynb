```python
import plotly.express as px
import pandas as pd
```

```python
df= px.data.gapminder()
df
```

```python
px.bar(df,x= 'country' ,y='lifeExp')
```

```python
filt = df['continent'] =='Asia' 
df_f = df.loc[filt]
df_f
```

```python
fig  = px.bar(df_f,x= 'country' ,y='lifeExp',color = 'country' ,animation_frame = 'year')

fig.update_layout(
    title='Life expectancy over the years',
    xaxis_title='Country',
    yaxis_title = 'Life Expectancy', 
)
```

```python
filt = (df['continent'] =='Asia')  & (df['year'] == 2007) 
df_f = df.loc[filt]
df_f
fig  = px.bar(df_f,x= 'country' ,y='lifeExp',color = 'country' ,animation_frame = 'year')

fig.update_layout(
    title='Life expectancy over the years',
    xaxis_title='Country',
    yaxis_title = 'Life Expectancy', 
)
```
