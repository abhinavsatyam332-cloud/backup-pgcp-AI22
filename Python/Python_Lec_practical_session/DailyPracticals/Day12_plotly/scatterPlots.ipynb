```python
import plotly.express as px
import pandas as pd
import seaborn as sns

```

```python
df = sns.load_dataset('diamonds')
df
```

```python
sample= df.sample(200)
```

```python
px.scatter(sample, x='carat',y ='price',color = 'cut')
```

```python
px.scatter(sample, x='carat',y ='price',color = 'cut',size = 'depth')
```

```python
px.scatter(sample, x='carat',y ='price',color = 'cut',symbol = 'depth')
```
