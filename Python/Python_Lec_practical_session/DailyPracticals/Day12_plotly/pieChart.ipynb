```python
import plotly.express as px
import pandas as pd
```

```python
df = px.data.tips()
df
```

```python
fig = px.pie(df, names = 'day'  ,values= 'total_bill')
fig
```

```python
fig = px.pie(df, names = 'day'  ,values= 'total_bill',hole = 0.5)
fig
```

```python
fig = px.pie(df, names = 'day'  ,values= 'total_bill',hole = 0.5)
fig

fig.update_layout(
    title='Life expectancy over the years',
    xaxis_title='Country',
    yaxis_title = 'Life Expectancy', 
)
```

```python
fig = px.pie(df, names = 'day'  ,values= 'total_bill',hole = 0.5, template = 'seaborn')
fig

fig.update_layout(
    title='Life expectancy over the years',
    xaxis_title='Country',
    yaxis_title = 'Life Expectancy', 
)
```
