Q1: What is the primary goal of deep learning?

-  B. To create shallow models
-  C. To learn hierarchical representations of data
-  D. To replace hardware components with software

Q2: Which of the following is a key characteristic of neural networks?
-  A. They can only process structured data
-  B. They are linear models
-  C. They consist of layers of interconnected nodes (neurons)
-  D. They cannot be used for regression problems


Answer: -  C. They consist of layers of interconnected nodes (neurons)

Q3: Which function ensures non-linearity in a neural network?
-  A. Dot product
-  B. Activation function
-  C. Weight initialization
-  D. Loss function

#### 4. Cost Function
Q10: What does the cost function measure in a neural network?
-  A. The number of layers in the network
-  B. The difference between predicted and actual values
-  C. The activation of neurons
-  D. The weight initialization strategy


Answer: -  B. The difference between predicted and actual values

Q11: What is the cost function typically used in logistic regression?
-  A. Mean Squared Error
-  B. Hinge Loss
-  C. Cross-Entropy Loss
-  D. Absolute Error Loss


Answer: -  C. Cross-Entropy Loss

Q12: In a binary classification task, what happens to the cost function if the predicted probability is very close to 0 for the actual positive class?
-  A. The cost is zero
-  B. The cost approaches infinity
-  C. The cost becomes negative
-  D. The cost does not change


Answer: -  B. The cost approaches infinity

#### 5. Activation
Q13: Which activation function is defined as f(x)=max(0,x)?
-  A. Sigmoid
-  B. ReLU
-  C. Tanh
-  D. Softmax


Answer: -  B. ReLU

Q14: Why is the ReLU activation function preferred in hidden layers?
-  A. It avoids the vanishing gradient problem
-  B. It is computationally expensive
-  C. It outputs values between -1 and 1
-  D. It reduces overfitting


Answer: -  A. It avoids the vanishing gradient problem

Q15: Which activation function is most suitable for the output layer in a multi-class classification problem?
-  A. Sigmoid
-  B. Tanh
-  C. Softmax
-  D. ReLU


Answer: -  C. Softmax

#### 6. Gradient Descent for Logistic Regression
Q16: What is the primary purpose of gradient descent?
-  A. To initialize weights
-  B. To find the global minimum of the cost function
-  C. To increase the learning rate
-  D. To compute activation values


Answer: -  B. To find the global minimum of the cost function

Q17: What determines the step size in gradient descent?
-  A. Cost function
-  B. Learning rate
-  C. Bias term
-  D. Regularization parameter


Answer: -  B. Learning rate

Q18: In stochastic gradient descent, what is the main difference from batch gradient descent?
-  A. It uses the entire dataset for each step
-  B. It updates weights after every single sample
-  C. It computes the exact gradient of the cost function
-  D. It increases the learning rate over time


Answer: -  B. It updates weights after every single sample

#### 7. Back-propagation
Q19: What does back-propagation compute?
-  A. Forward pass predictions
-  B. Gradients of the cost function with respect to weights
-  C. The output of the network
-  D. The activation of neurons in the input layer


Answer: -  B. Gradients of the cost function with respect to weights

Q20: Why is the chain rule essential in back-propagation?
-  A. To initialize weights in neural networks
-  B. To propagate error gradients backward through layers
-  C. To perform forward propagation
-  D. To normalize input data


Answer: -  B. To propagate error gradients backward through layers

Q21: In backp-ropagation, why do deep networks sometimes struggle to update weights effectively?
-  A. Overfitting
-  B. Vanishing or exploding gradients
-  C. Insufficient training data
-  D. Lack of dropout layers


Answer: -  B. Vanishing or exploding gradients
